```python
#Matplotlib
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,100)
y=np.sin(x)
z=np.cos(x)
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,z)
plt.show()
```

```python
plt.xlabel('angle')
plt.ylabel('value')
plt.title('sine wave')
plt.plot(x,y)
plt.show()
```

```python
#bar graph
fig = plt.figure()
ax=fig.add_axes([0,0,1,1])
languages=['c','c++','python','java']
students=[23,17,35,30]
ax.bar(languages,students)
plt.xlabel('languages')
plt.ylabel('number of students')
plt.title('students enrolled')
plt.show()
```

```python
#using pie chart
fig = plt.figure()
ax=fig.add_axes([0,0,1,1])
languages=['c','c++','python','java']
students=[23,17,35,30]
ax.pie(students,labels=languages,autopct='%1.2f%%')
plt.show()
```

```python
#scatter plot
x=[5,7,8,7,2,17,2,9,4,11,12,9,6]
y=[99,86,87,88,111,86,103,87,94,78,77,85,86]
plt.scatter(x,y)
plt.show()
```
